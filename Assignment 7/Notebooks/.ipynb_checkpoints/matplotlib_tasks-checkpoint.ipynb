```python
# Task 6: Line Plot
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Name": ["Chris", "Eska", "Kpv", "Mans", "GPT"],
    "Age": [102, 21, 20, 22, 2],
    "Score": [88, 74, 91, 83, 95]
})
```

```python
plt.plot(df["Name"], df["Score"], marker="o")
plt.xlabel("Student")
plt.ylabel("Score")
plt.title("Student Scores (Line Plot)")
plt.xticks(rotation=30)
plt.grid(True)
plt.show()
```

```python
# Task 7: Bar Chart
```

```python
plt.bar(df["Name"], df["Score"])
plt.xlabel("Student")
plt.ylabel("Score")
plt.title("Student Scores (Bar Chart)")
plt.xticks(rotation=30)
plt.show()
```
