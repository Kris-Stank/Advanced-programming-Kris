```python
# Task 3: Create a DataFrame
```

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Chris", "Mansyr", "Eska", "Snanzhar", "GPT"],
    "Age": [23, 40, 15, 20, 2],
    "Score": [100, 74, 91, 83, 95]
})

df
```

```python
# Task 4: Data Exploration
```

```python
print("First 3 rows:")
display(df.head(3))

print("Describe:")
display(df.describe())

avg_score = df["Score"].mean()
print("Average score:", avg_score)
```

```python
# Task 5: Data Filtering
```

```python
print("Age > 20:")
display(df[df["Age"] > 20])

print("Score > 80:")
display(df[df["Score"] > 80])
```
